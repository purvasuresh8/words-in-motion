Spanish

```python
from datasets import load_dataset
import pandas as pd

dataset = load_dataset(
    "sentence-transformers/parallel-sentences-opensubtitles",
    "en-es",
    split="train",
    streaming=True
)

rows = []

for i, item in enumerate(dataset):
    rows.append(item)
    if i >= 4999:
        break

df = pd.DataFrame(rows)

print(df.shape)
print(df.columns.tolist())
print(df.head())
```

```python
import string
import pandas as pd

# Remove missing and empty sentence pairs
df = df.dropna(subset=["english", "non_english"]).copy()

df["english"] = df["english"].astype(str).str.strip()
df["non_english"] = df["non_english"].astype(str).str.strip()

df = df[
    (df["english"] != "") &
    (df["non_english"] != "")
].copy()

# Count characters, excluding spaces and punctuation
def count_characters(text):
    return sum(
        1 for char in text
        if not char.isspace() and char not in string.punctuation
    )

# Count words using whitespace separation
def count_words(text):
    return len(text.split())

df["english_chars"] = df["english"].apply(count_characters)
df["translated_chars"] = df["non_english"].apply(count_characters)

df["english_words"] = df["english"].apply(count_words)
df["translated_words"] = df["non_english"].apply(count_words)

# Remove sentences that cannot produce a character ratio
df = df[df["english_chars"] > 0].copy()

df["character_expansion_ratio"] = (
    df["translated_chars"] / df["english_chars"]
)

df["word_expansion_ratio"] = (
    df["translated_words"] / df["english_words"]
)

print("Usable sentence pairs:", len(df))
display(df.head(10))
```

```python
df.to_csv("words_in_motion_spanish.csv", index=False)

print("CSV exported successfully!")
```
